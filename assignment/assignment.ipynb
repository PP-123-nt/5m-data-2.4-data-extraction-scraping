```python
import requests
import time
from bs4 import BeautifulSoup

def parse_and_extract_rows(soup: BeautifulSoup):
    """
    Extract table rows from the parsed HTML.
    (Your original function remains unchanged)
    """
    header = soup.find('tr')
    if not header:
        return
    headers = [th.text.strip() for th in header.find_all('th')]
    teams = soup.find_all('tr', 'team')
    for team in teams:
        row_dict = {}
        for header, col in zip(headers, team.find_all('td')):
            row_dict[header] = col.text.strip()
        yield row_dict

# --- New Scraping Logic ---
base_url = "https://www.scrapethissite.com/pages/forms/"
page_num = 1

# Use a while loop to keep scraping until there is no "Next" button
while True:
    print(f"Scraping page {page_num}...")
    
    # Construct the URL for the current page
    url = f"{base_url}?page_num={page_num}"
    
    # Fetch the page
    response = requests.get(url)
    soup = BeautifulSoup(response.content, 'html.parser')
    
    # 1. Extract data from the current page
    for row in parse_and_extract_rows(soup):
        print(row) # You can save this to a list, CSV, database, etc.
        
    # 2. Look for the '»' (Next) button
    # On scrapethissite.com, the next button has an aria-label="Next"
    next_button = soup.find('a', attrs={'aria-label': 'Next'})
    
    # 3. Determine if we should continue
    if next_button:
        page_num += 1
        time.sleep(1) # Always be polite and add a delay between requests
    else:
        print("No more pages found. Scraping complete!")
        break # Exit the while loop
```

✏️✏️✏️ **How it works:** ✏️✏️✏️

1. Initialization: We start at `page_num = 1`.

2. The `while True` loop: This creates an infinite loop that we will manually `break` out of when we are done.

3. Dynamic URLs: The URL is built using an f-string: `f"{base_url}?page_num={page_num}"`.

4. Finding the `»` button: `soup.find('a', attrs={'aria-label': 'Next'})` searches the HTML for a link `<a>` tag that acts as the next button.

5. The Break Condition:

    - If `next_button` is found, it means there are more pages. We increment page_num by 1, wait 1 second (`time.sleep(1)`), and the loop runs again.

    - If `next_button` is `None`, it means we have reached the last page. The `else` block triggers, prints a message, and the `break` statement stops the loop.

```python
# print the number of rows extracted

import requests
import time
from bs4 import BeautifulSoup

def parse_and_extract_rows(soup: BeautifulSoup):
    """
    Extract table rows from the parsed HTML.
    """
    header = soup.find('tr')
    if not header:
        return
    headers = [th.text.strip() for th in header.find_all('th')]
    teams = soup.find_all('tr', 'team')
    for team in teams:
        row_dict = {}
        for header, col in zip(headers, team.find_all('td')):
            row_dict[header] = col.text.strip()
        yield row_dict

# --- New Scraping Logic ---
base_url = "https://www.scrapethissite.com/pages/forms/"
page_num = 1
total_rows = 0  # Initialize the row counter

while True:
    print(f"Scraping page {page_num}...")
    
    url = f"{base_url}?page_num={page_num}"
    response = requests.get(url)
    soup = BeautifulSoup(response.content, 'html.parser')
    
    # Extract data from the current page
    for row in parse_and_extract_rows(soup):
        # print(row) # You can uncomment this if you want to see every row
        total_rows += 1  # Increment the counter for each row extracted
        
    # Look for the '»' (Next) button
    next_button = soup.find('a', attrs={'aria-label': 'Next'})
    
    if next_button:
        page_num += 1
        time.sleep(1) 
    else:
        print("No more pages found. Scraping complete!")
        break 

# Print the final count
print(f"\nTotal number of rows extracted across all pages: {total_rows}")
```
